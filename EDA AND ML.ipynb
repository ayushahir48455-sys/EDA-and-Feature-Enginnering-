```python
# here we import importatnt libraray
import pandas as pd
import numpy as np

# Loading the data in DataFrame
df = pd.read_csv('Aayush.csv')
df.head()
```

```python
#Missing Values Handling
df.isnull().sum().sum() # total  number of missing values in the whole data set

df.isnull().sum()# Detects total missing (NaN) values in column 

(df.isnull().sum()/df.size)*100 # its gives the percentage of data is missing coloumn wise

print((df.isnull().sum().sum()/df.size)*100) # total percentage of null value present in the data set

# if you want to find notnull values then replace the null to notnull
```

```python
df.describe()
```

```python
df.value_counts()
```

```python
df.info()
```

```python
# covert data type of phone number and remove the null value

df['Phone'] = pd.to_numeric(df['Phone'], errors='coerce')
df['Phone'] = df['Phone'].fillna(df['Phone'].median())
df.head()
```

```python

# covert gender coloum to lower case then replace female to f
df['Gender'] = df['Gender'].str.lower()
df['Gender'] = df['Gender'].replace({'female' : 'f'})
df.head()
```

```python
#convert to data time formate
df['Join_Date'] = df['Join_Date'].ffill()
```

```python
# fill email using notproved
df['Email'] = df['Email'].fillna('not provided')
df.head()
```

```python
#fill city column
df['City'] = df['City'].fillna(df['City'].mode()[0])
df.head()
```

```python
# fill nan value
df['Salary'] = df['Salary'].fillna(df['Salary'].median())
df.head()
```

```python
df['Age'] = pd.to_numeric(df['Age'], errors='coerce')
```

```python
print(df['Age'].dtype)
print(df['Age'].unique())
```

```python
# ✅ Step 1️⃣ Detect Outliers (Using IQR Method – Best Practice)
Q1 = df['Age'].quantile(0.25)
Q3 = df['Age'].quantile(0.75)
IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

print("Lower Limit:", lower_limit)
print("Upper Limit:", upper_limit)

# ✅ Step 2️⃣ Find Outliers
# used to see the outlier
outliers = df[(df['Age'] < lower_limit) | (df['Age'] > upper_limit)]
print("outliers are :",outliers[['Age']])

#✅ Step 3️⃣ Handle Outliers

# Option 1: Remove Outliers (Most Common)
#df = df[(df['Age'] >= lower_limit) & (df['Age'] <= upper_limit)]
#print(df)

#Option 2: Replace Outliers with Median (Safer)
median_age = df['Age'].median()


df.loc[(df['Age'] < lower_limit) | (df['Age'] > upper_limit), 'Age'] = median_age
df.head(3)
```

```python
min_range = df["Salary"].mean() - (3*df["Salary"].std())
max_range = df["Salary"].mean() + (3*df["Salary"].std())
min_range,max_range
```

```python
#Using Z-score
z_score = (df["Age"]-df["Age"].mean())/(df["Age"].std())
df["z_score"] = z_score
df[df["z_score"]<3].head(3)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Drop missing ratings
ratings = df['Rating'].dropna()

# Categorize ratings into color groups
def get_color(r):
    if r > 4.5:
        return 'green'
    elif r > 4 and r <= 4.5:
        return 'red'
    elif r > 3 and r < 4:
        return 'blue'
    else:
        return 'gray'  # fallback

color_labels = ratings.apply(get_color)

# Count ratings by category
color_counts = color_labels.value_counts()

print("Total ratings by color category:")
print(color_counts)

# Map labels to actual colors
color_map = {
    'green': 'green',
    'red': 'red',
    'blue': 'blue',
    'gray': 'gray'
}

# Plot pie chart
plt.figure(figsize=(6,6))
plt.pie(color_counts,
        labels=[f"{label} ({count})" for label, count in zip(color_counts.index, color_counts)],
        colors=[color_map[label] for label in color_counts.index],
        autopct='%1.1f%%',
        startangle=90)
plt.title("Ratings Distribution by Color Category")
plt.show()
```

# This is another Data set and we perform Feature engineering

```python
import numpy as np 
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings('ignore')
df = pd.read_csv("Project_1.csv")
df.head(5)
```

```python
# ploting histogram ,Unovariate analysis
plt.figure(figsize = (4,2))
sns.histplot(df['price'],bins = 50,kde = True)
```

```python
#Box plot
plt.figure(figsize = (4,2))
sns.boxplot(data = df,x = df['model'],y = df['price'])
plt.xticks(rotation = 90)
```

```python
#here we just take our all column excet price in x
x = df.drop(columns = ['price'],axis = 1)
y = df['price']
```

```python
# here we encode the 3 columns to use there

x_encode = pd.get_dummies(x,columns = ['model','transmission','fuelType'],drop_first = True)
x_encode.head(3)
```

```python
# covertind all the encoded data into numeric
x_encode = x_encode.astype(int)
x_encode.head(3)
```

```python
#A Label Encoder converts categorical values into numeric codes so machine learning models can process them.

from sklearn.preprocessing import LabelEncoder
encode = LabelEncoder()

columns = ['model','transmission','fuelType']
xlable = x

for i in columns:
   xlable[i]=  encode.fit_transform(xlable[i])

xlable
```

```python
#standatdscaler the columns in -1 and 1
from sklearn.preprocessing import StandardScaler

num_col = ['year','mileage','tax','mpg','engineSize']
scaler = StandardScaler()

x_encode[num_col] = scaler.fit_transform(x_encode[num_col])
x_encode.head(5)
```

```python
# using oneHotEncoder

xlable[['year','model','transmission','mileage','fuelType','tax','mpg','engineSize']] = scaler.fit_transform(xlable[['model','year','transmission','mileage','tax','mpg','fuelType','engineSize']])
xlable
```

```python
#Here we build model for your new data 

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error,mean_squared_error,r2_score
```

```python
x_train,x_test,y_train,y_test = train_test_split(x_encode,y,test_size = 0.33,random_state = 42)
lr = LinearRegression()
lr.fit(x_train,y_train)
```

```python
y_pred = lr.predict(x_test)

# use r2
r2 = r2_score(y_test,y_pred)

n = x_test.shape[0]
p = x_test.shape[1]
adjusted_r2 = 1-((1-r2)*(n-1)/(n-p-1))
print(adjusted_r2)
```

```python
# building model on same xlable data
x_train,x_test,y_train,y_test = train_test_split(xlable,y,test_size = 0.33,random_state = 42)
lr2 = LinearRegression()
lr2.fit(x_train,y_train)
```

```python
y_pred = lr2.predict(x_test)

# use r2
r2 = r2_score(y_test,y_pred)

n = x_test.shape[0]
p = x_test.shape[1]
adjusted_r2 = 1-((1-r2)*(n-1)/(n-p-1))
print(adjusted_r2)
```
